```python
import pandas as pd
import numpy as np
import os

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
from sklearn.cluster import KMeans

import matplotlib.pyplot as plt
%matplotlib inline

taxi_data = pd.read_csv('./data/sample.csv', encoding='euc-kr')
taxi_data = taxi_data[taxi_data['승차X좌표'] != 0.0]

# 시각 분리: int type
taxi_data['시각'] = taxi_data['승차시간'].apply(lambda x: int(x[11:13]+x[14:16]+x[17:])) # int 적용?
# 요일 원핫인코딩
day_encoding = pd.get_dummies(taxi_data['요일'])
day_encoding = day_encoding.astype(int)
taxi_data = taxi_data.drop('요일', axis=1)
taxi_data = taxi_data.join(day_encoding)
taxi_data = taxi_data.drop(['승차시간', '차량번호'], axis=1)

# ==================================================================
cluster_data = taxi_data[['승차X좌표', '승차Y좌표']]

kmeans = KMeans(n_clusters=9, init='random')
kmeans.fit(cluster_data)
labels = kmeans.labels_

for i, label in enumerate(labels):
    if label == 0:
        color = 'blue'
    elif label == 1:
        color = 'red'
    elif label == 2:
        color == 'green'
    elif label == 3:
        color == 'yellow'
    elif label == 4:
        color == 'orange'
    elif label == 5:
        color == 'purple'
    elif label == 6:
        color == 'pink'
    elif label == 7:
        color == 'black'
    else:
        color == 'sky'
    plt.scatter(cluster_data[i]['승차X좌표'], cluster_data[i]['승차Y좌표'], color=color)

plt.show()

# ==================================================================

# 학습, 테스트 데이터 분리
X = taxi_data[taxi_data.columns.difference(['승차X좌표', '승차Y좌표'])]
y = taxi_data[['승차X좌표', '승차Y좌표']]
train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.3, random_state=42)




```
